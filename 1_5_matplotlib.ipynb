# 📊 Introduction to Matplotlib

## What is Matplotlib?

**Matplotlib** is a Python library used for **data visualization** — it helps us convert numerical data into **graphs and plots** so that we can *see patterns, trends, and relationships* in the data instead of only reading numbers.

In simple words:

> **Matplotlib helps you "see" your data.**

It is widely used in:

* Data Science
* Machine Learning
* Deep Learning
* Scientific computing
* Research and reports

Anytime you want to understand data better, debug a model, or present results visually — Matplotlib is used.

---

## Why do we need Matplotlib?

When working with data or training ML models:

* Numbers alone are hard to interpret.
* Visuals make understanding **faster and more intuitive**.

Matplotlib allows you to:

* Plot **lines, points, bars, histograms, images**, etc.
* Visualize **training loss**, accuracy curves, distributions, comparisons.
* Detect **outliers, trends, and errors** in data.

Example use-cases:

* Plotting loss vs epochs while training a neural network.
* Visualizing how two variables are related.
* Showing the distribution of a dataset.
* Comparing model performances.

---

## Where Matplotlib fits in the ML / PyTorch workflow

Typical workflow:

1. Load & prepare data
2. Explore data visually (**Matplotlib here**)
3. Build model (PyTorch)
4. Train model
5. Visualize results (**Matplotlib again**)
6. Analyze and improve

So Matplotlib is mainly used in:

* **Data exploration**
* **Model evaluation**
* **Result presentation**

---

## What Matplotlib actually does

Matplotlib:

* Takes numerical data (lists, NumPy arrays, PyTorch tensors)
* Converts them into visual representations
* Renders them as plots inside notebooks or windows

It acts as a **bridge between data and human understanding**.

---

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

x_axis = [1,2,3,4,5,6]

y_axis = [1,4,9,16,25,36]

plt.plot(x_axis, y_axis)
# If you want grid in the Graph
plt.grid()
plt.show()
````

`pyplot` is a submodule of Matplotlib that provides a simple, MATLAB-like interface for creating plots.

It acts as a state-based plotting API — meaning it keeps track of:

* The current figure

* The current axes

* The current plot

So you can make plots with very little code, without manually creating figure and axes objects.

## 📈 Univariate Analysis

### What is Univariate?

**Univariate** means **“one variable”**.

So **univariate analysis** is the analysis of **a single variable at a time**, without considering any relationship with other variables.

> **Uni = one, Variate = variable → Univariate = one variable**

---

### What is Univariate Analysis?

Univariate analysis focuses on:

* Understanding the **distribution**
* Central tendency (mean, median, mode)
* Spread (variance, standard deviation)
* Shape (skewness, outliers)

of a single feature or column in a dataset.

---

### Why is Univariate analysis important?

Before building models, you must:

* Understand your data
* Detect outliers
* See if values are skewed
* Check ranges and distributions

Univariate analysis helps answer:

* What values does this variable take?
* How often does each value occur?
* Is it normally distributed or skewed?
* Are there extreme values?

---

### Examples of univariate plots

Typical plots used for univariate analysis:

| Plot      | Purpose                                 |
| --------- | --------------------------------------- |
| Histogram | Distribution of numerical data          |
| Box plot  | Outliers and spread                     |
| Bar chart | Frequency of categorical values         |
| Line plot | Trend over index/time (single variable) |

---

### Univariate vs Multivariate

| Type         | Variables | Example                   |
| ------------ | --------- | ------------------------- |
| Univariate   | 1         | Distribution of age       |
| Bivariate    | 2         | Height vs weight          |
| Multivariate | 3+        | Price vs size vs location |

## Why `pyplot` exists

Matplotlib internally works with:

* Figures
* Axes
* Artists (lines, texts, patches, etc.)

That system is powerful but verbose.

`pyplot` simplifies this by:

* Automatically creating figures and axes
* Managing plot state for you
* Letting you focus only on *what* to plot, not *how to build the plot object structure*

---

## What `pyplot` controls

`pyplot` handles:

| Component     | Role                                      |
| ------------- | ----------------------------------------- |
| Figure        | The whole plotting window                 |
| Axes          | The actual plot area inside the figure    |
| Plot elements | Lines, points, bars, text, legends, grids |

You usually interact with all of this indirectly through `plt`.

---

## Common `pyplot` functions

### Creating plots

* `plt.plot()` → line plots
* `plt.scatter()` → scatter plots
* `plt.bar()` → bar charts
* `plt.hist()` → histograms
* `plt.imshow()` → images

### Labels and titles

* `plt.title()` → plot title
* `plt.xlabel()` → x-axis label
* `plt.ylabel()` → y-axis label
* `plt.legend()` → legend
* `plt.grid()` → grid lines

### Figure control

* `plt.figure()` → create new figure
* `plt.show()` → render the plot
* `plt.savefig()` → save plot to file
* `plt.clf()` → clear current figure

---

## `pyplot` vs Object-Oriented API

Matplotlib has two styles:

### 1️⃣ `pyplot` (state-based, simple)

Good for:

* Quick plots
* Notebooks
* Exploratory analysis

```python
plt.plot(x, y)
plt.show()
```

### 2️⃣ Object-Oriented API (explicit control)

Good for:

* Complex layouts
* Multiple subplots
* Production-quality figures

```python
fig, ax = plt.subplots()
ax.plot(x, y)
```

---

## When to use `pyplot`

Use `pyplot` when:

* You want fast visualization
* You are exploring data
* You are working in Colab / Jupyter
* You don’t need very complex layout control

Use the object-oriented API when:

* You need multiple plots
* You want fine-grained control
* You are building dashboards or publications
